```python
%sql
CREATE OR REPLACE VIEW retail_p.retail_semantic.retail_metrics
WITH METRICS
LANGUAGE YAML
AS $$
version: 1.1
source: retail_p.retail_gold.fact_sales
comment: Retail sales metrics combining transactions with product, customer, and calendar dimensions
joins:
  - name: product
    source: retail_p.retail_gold.dim_product
    on: source.product_id = product.product_id
  - name: customer
    source: retail_p.retail_gold.dim_customer
    on: source.customer_id = customer.customer_id
  - name: calendar
    source: retail_p.retail_gold.dim_calendar
    on: source.transaction_date = calendar.date
dimensions:
  - name: product_category
    expr: product.category
    display_name: Product Category
    comment: Product category from product dimension
  - name: customer_name
    expr: customer.customer_name
    display_name: Customer Name
    comment: Customer name from customer dimension
  - name: customer_industry
    expr: customer.industry
    display_name: Customer Industry
    comment: Customer industry from customer dimension
  - name: billing_country
    expr: customer.billing_country
    display_name: Billing Country
    comment: Customer billing country from customer dimension
  - name: billing_state
    expr: customer.billing_state
    display_name: Billing State
    comment: Customer billing state from customer dimension
  - name: customer_type
    expr: source.stage_name
    display_name: Customer Type
    comment: Customer stage in sales pipeline
  - name: sales_channel
    expr: source.sales_channel
    display_name: Sales Channel
    comment: Channel through which sale was made
  - name: payment_mode
    expr: source.payment_mode
    display_name: Payment Mode
    comment: Method of payment used
  - name: year
    expr: calendar.year
    display_name: Year
    comment: Year when transaction occurred
  - name: quarter
    expr: calendar.quarter
    display_name: Quarter
    comment: Quarter when transaction occurred
  - name: year_quarter
    expr: calendar.year_quarter
    display_name: Year Quarter
    comment: Year and quarter combined
  - name: year_month
    expr: calendar.year_month
    display_name: Year Month
    comment: Year and month combined
  - name: month
    expr: calendar.month_name
    display_name: Month
    comment: Month name when transaction occurred
  - name: day_of_week
    expr: calendar.day_of_week_name
    display_name: Day of Week
    comment: Day of week when transaction occurred
measures:
  - name: total_revenue
    expr: SUM(amount)
    display_name: Total Revenue
    comment: Total sales revenue
    format:
      type: currency
      currency_code: USD
      decimal_places:
        type: exact
        places: 2
  - name: total_quantity
    expr: SUM(quantity)
    display_name: Total Quantity
    comment: Total quantity of products sold
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
  - name: total_discount
    expr: SUM(discount_amount)
    display_name: Total Discount
    comment: Total discount amount given
    format:
      type: currency
      currency_code: USD
      decimal_places:
        type: exact
        places: 2
  - name: transaction_count
    expr: COUNT(DISTINCT transaction_id)
    display_name: Transaction Count
    comment: Number of unique transactions
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
  - name: avg_transaction_value
    expr: SUM(amount) / COUNT(DISTINCT transaction_id)
    display_name: Average Transaction Value
    comment: Average value per transaction
    format:
      type: currency
      currency_code: USD
      decimal_places:
        type: exact
        places: 2
$$
```
